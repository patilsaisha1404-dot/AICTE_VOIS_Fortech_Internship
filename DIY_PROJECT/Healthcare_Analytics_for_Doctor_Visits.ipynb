```python
# ============================================================
# HEALTHCARE ANALYTICS FOR DOCTOR VISITS
# Complete EDA + Data Visualization
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_style("whitegrid")

# ------------------------------------------------------------
# 2. LOAD DATASET
# ------------------------------------------------------------

# For Google Colab
df = pd.read_csv(
    "1776250375-P2-Healthcare Analytics for Doctor Visits (1).csv"
)
```

```python
# 3. DISPLAY FIRST 5 ROWS
# ------------------------------------------------------------

print("First 5 rows:")
display(df.head())


# ------------------------------------------------------------
# 4. DISPLAY LAST 5 ROWS
# ------------------------------------------------------------

print("Last 5 rows:")
display(df.tail())
```

```python

print("Number of rows and columns:")
print(df.shape)


# ------------------------------------------------------------
# 6. DATASET INFORMATION
# ------------------------------------------------------------

print("\nDataset Information:")
df.info()
```

```python

print("\nStatistical Information:")
display(df.describe())


# ------------------------------------------------------------
# 8. CHECK NULL VALUES
# ------------------------------------------------------------

print("\nNull Values:")
print(df.isnull().sum())


# ------------------------------------------------------------
# 9. CHECK DUPLICATE VALUES
# ------------------------------------------------------------

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="gender",
    hue="gender",
    legend=False
)

plt.title("Distribution of Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="age",
    bins=20,
    kde=True
)

plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="visits",
    bins=range(int(df["visits"].max()) + 2),
    discrete=True
)

plt.title("Distribution of Doctor Visits")
plt.xlabel("Number of Doctor Visits")
plt.ylabel("Number of Patients")
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="illness",
    hue="illness",
    legend=False
)

plt.title("Distribution of Illness Score")
plt.xlabel("Illness Score")
plt.ylabel("Number of Patients")
plt.show()
```

```python

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="health",
    hue="health",
    legend=False
)

plt.title("Distribution of Health Status")
plt.xlabel("Health Status")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="income",
    bins=20,
    kde=True
)

plt.title("Distribution of Income")
plt.xlabel("Income")
plt.ylabel("Frequency")
plt.show()
```

```python
gender_analysis = (
    df.groupby("gender")["visits"]
      .agg(["count", "mean", "median"])
      .reset_index()
)

print("Doctor Visits by Gender:")
display(gender_analysis)


plt.figure(figsize=(7, 5))

sns.barplot(
    data=gender_analysis,
    x="gender",
    y="mean"
)

plt.title("Average Doctor Visits by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Number of Visits")
plt.show()
```

```python

illness_analysis = (
    df.groupby("illness")["visits"]
      .agg(["count", "mean"])
      .reset_index()
)

print("Doctor Visits by Illness Score:")
display(illness_analysis)


plt.figure(figsize=(8, 5))

sns.lineplot(
    data=illness_analysis,
    x="illness",
    y="mean",
    marker="o"
)

plt.title("Illness Score vs Average Doctor Visits")
plt.xlabel("Illness Score")
plt.ylabel("Average Doctor Visits")
plt.show()
```

```python
gender_health = pd.crosstab(
    df["gender"],
    df["health"]
)

print("Gender vs Health Status:")
display(gender_health)


gender_health.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Gender vs Health Status")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.xticks(rotation=0)
plt.legend(title="Health")
plt.show()
```

```python
insurance_columns = [
    "private",
    "freepoor",
    "freerepat"
]

insurance_counts = []

for col in insurance_columns:
    insurance_counts.append(
        df[col].value_counts().get("yes", 0)
    )

plt.figure(figsize=(8, 5))

plt.bar(
    insurance_columns,
    insurance_counts
)

plt.title("Number of Patients with Different Insurance/Support")
plt.xlabel("Insurance / Support Type")
plt.ylabel("Number of Patients")
plt.show()
```

```python
df_numeric = df.copy()

# Convert yes/no columns
binary_columns = [
    "private",
    "freepoor",
    "freerepat",
    "nchronic",
    "lchronic"
]

for col in binary_columns:
    df_numeric[col] = df_numeric[col].map({
        "yes": 1,
        "no": 0
    })

# Convert gender
df_numeric["gender"] = df_numeric["gender"].map({
    "male": 1,
    "female": 0
})
```

```python
df_numeric["gender"] = df_numeric["gender"].map({
    "male": 1,
    "female": 0
})


# ------------------------------------------------------------
# 29. CORRELATION MATRIX
# ------------------------------------------------------------

correlation = df_numeric.corr(numeric_only=True)

print("Correlation Matrix:")
display(correlation)
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Healthcare Variables")
plt.show()
```

```python
age_visits = (
    df.groupby("age")["visits"]
      .mean()
      .reset_index()
)

plt.figure(figsize=(10, 5))

sns.lineplot(
    data=age_visits,
    x="age",
    y="visits",
    marker="o"
)

plt.title("Age vs Average Doctor Visits")
plt.xlabel("Age")
plt.ylabel("Average Doctor Visits")
plt.show()
```

```python
health_visits = (
    df.groupby("health")["visits"]
      .mean()
      .reset_index()
)

print("Average Visits by Health Status:")
display(health_visits)


plt.figure(figsize=(8, 5))

sns.barplot(
    data=health_visits,
    x="health",
    y="visits"
)

plt.title("Average Doctor Visits by Health Status")
plt.xlabel("Health Status")
plt.ylabel("Average Doctor Visits")
plt.show()
```
