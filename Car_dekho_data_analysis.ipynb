```python
import pandas as pd

# Load the dataset CSV file
df = pd.read_csv("1776311302-P3-Car Market Trends Analysis with Car Dekho Data.csv")

# Display the first few rows to confirm it loaded correctly
df.head()
```

```python
start_year = df['Year'].min()
end_year = df['Year'].max()
print(f"Vehicles present from year {start_year} to {end_year}")
```

```python
lowest_price = df['Selling_Price'].min()
print(f"Lowest selling price: {lowest_price} Lakhs")
```

```python
highest_price = df['Selling_Price'].max()
print(f"Highest selling price: {highest_price} Lakhs")
```

```python
total_records = len(df)
print(f"Total records: {total_records}")
```

```python
missing_records = df.isnull().sum()
print("Missing values per column:\n", missing_records)
```

```python
unique_vehicles = df['Car_Name'].nunique()
print(f"Number of unique vehicles: {unique_vehicles}")
```

```python
most_sold_vehicle = df['Car_Name'].value_counts().idxmax()
count = df['Car_Name'].value_counts().max()
print(f"Most sold vehicle: {most_sold_vehicle} ({count} occurrences)")
```

```python
cng_count = (df['Fuel_Type'] == 'CNG').sum()
print(f"Number of CNG vehicles: {cng_count}")
```

```python
individual_sellers = (df['Seller_Type'] == 'Individual').sum()
print(f"Vehicles sold by Individuals: {individual_sellers}")
```

```python
automatic_count = (df['Transmission'] == 'Automatic').sum()
print(f"Automatic transmission vehicles: {automatic_count}")
```

```python
single_owner_count = (df['Owner'] == 0).sum()
print(f"Single person owned vehicles: {single_owner_count}")
```

```python
df['Cost_Depreciation'] = df['Present_Price'] - df['Selling_Price']

most_depreciated = df.loc[df['Cost_Depreciation'].idxmax()]
least_depreciated = df.loc[df['Cost_Depreciation'].idxmin()]

print(f"Most Depreciated: {most_depreciated['Car_Name']} ({most_depreciated['Cost_Depreciation']:.2f} Lakhs)")
print(f"Least Depreciated: {least_depreciated['Car_Name']} ({least_depreciated['Cost_Depreciation']:.2f} Lakhs)")
```

```python
df['Brand'] = df['Car_Name'].apply(lambda x: x.split()[0])
avg_dep_by_brand = df.groupby('Brand')['Cost_Depreciation'].mean().sort_values()
print("Brands least affected by depreciation:\n", avg_dep_by_brand.head(5))
```

```python
corr_matrix = df[['Cost_Depreciation', 'Present_Price', 'Year', 'Kms_Driven']].corr()
print("Correlation with Cost Depreciation:\n", corr_matrix['Cost_Depreciation'])
```

```python
print("Correlation of Selling Price with Year:", df['Selling_Price'].corr(df['Year']))
print("Correlation of Selling Price with Kms Driven:", df['Selling_Price'].corr(df['Kms_Driven']))
```

```python
newest_vehicles = df[df['Year'] > 2014]
print(f"Number of vehicles manufactured after 2014: {len(newest_vehicles)}")
print(newest_vehicles.head())
```

```python
bike_keywords = ['Royal Enfield', 'KTM', 'Bajaj', 'Yamaha', 'TVS', 'Hero', 'Activa', 'CB', 'Access', 'Discover', 'Mojo', 'Renegade', 'Hyosung']
is_bike = lambda name: any(k.lower() in name.lower() for k in bike_keywords)

bikes = df[df['Car_Name'].apply(is_bike)]
print(f"Total two-wheelers found: {len(bikes)}")
```

```python
oldest_bike = bikes.loc[bikes['Year'].idxmin()]
print(f"Oldest bike: {oldest_bike['Car_Name']} ({oldest_bike['Year']})")
```

```python
newest_bike = bikes.loc[bikes['Year'].idxmax()]
print(f"Newest bike: {newest_bike['Car_Name']} ({newest_bike['Year']})")
```

```python
most_sold_bike = bikes['Car_Name'].value_counts().idxmax()
print(f"Most sold bike: {most_sold_bike} ({bikes['Car_Name'].value_counts().max()} sales)")
```

```python
high_val_bikes = bikes[bikes['Selling_Price'] > 1.5]
print(high_val_bikes[['Car_Name', 'Year', 'Selling_Price', 'Present_Price', 'Kms_Driven']])
```

```python
cars = df[~df['Car_Name'].apply(is_bike)]
print(f"Total cars found: {len(cars)}")
```

```python
oldest_car = cars.loc[cars['Year'].idxmin()]
print(f"Oldest car: {oldest_car['Car_Name']} ({oldest_car['Year']})")
```

```python
newest_car = cars.loc[cars['Year'].idxmax()]
print(f"Newest car: {newest_car['Car_Name']} ({newest_car['Year']})")
```

```python
high_val_cars = cars[cars['Selling_Price'] > 15.0]
print(high_val_cars[['Car_Name', 'Year', 'Selling_Price', 'Present_Price', 'Kms_Driven', 'Transmission']])
```
